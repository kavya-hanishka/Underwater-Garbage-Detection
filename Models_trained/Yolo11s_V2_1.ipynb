```python
V21 = "/content/drive/MyDrive/material_version/V2.1"

class_names = [
    "rov",
    "plant",
    "animal_fish",
    "animal_starfish",
    "animal_shells",
    "animal_crab",
    "animal_eel",
    "animal_etc",
    "trash_etc",
    "trash_fabric",
    "trash_fishing_gear",
    "trash_metal",
    "trash_paper",
    "trash_plastic",
    "trash_rubber",
    "trash_wood"
]

yaml_path = os.path.join(V21, "data.yaml")

with open(yaml_path, "w") as f:
    f.write(f"path: {V21}\n")
    f.write("train: images/train\n")
    f.write("val: images/val\n")
    f.write("test: images/test\n")
    f.write("nc: 16\n")
    f.write("names:\n")

    for i, name in enumerate(class_names):
        f.write(f"  {i}: {name}\n")

print("Created:", yaml_path)
print("Exists:", os.path.exists(yaml_path))

with open(yaml_path, "r") as f:
    print(f.read())
```

```python
from ultralytics import YOLO

DATA = "/content/drive/MyDrive/material_version/V2.1/data.yaml"

model = YOLO("yolo11s.pt")

results = model.train(
    data=DATA,
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    project="/content/drive/MyDrive/material_version/yolo_runs",
    name="V2.1_yolo11s_baseline_640",
    pretrained=True,
    seed=42,
    workers=4
)
```
