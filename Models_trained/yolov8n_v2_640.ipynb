```python
import os
import json
import hashlib
import random
import shutil
import base64
import zlib
import io
from collections import defaultdict,Counter
from PIL import Image

# =========================
# PATHS
# =========================
BASE="/kaggle/input/datasets/hanishkavya/material-version/dataset"

TRAIN_SRC=os.path.join(BASE,"material_version/train")
VAL_SRC=os.path.join(BASE,"material_version/val")
IMG_SRC=os.path.join(BASE,"original_data/images")
ANN_SRC=os.path.join(BASE,"original_data/annotations")

OUT="/kaggle/working/yolo_v2"

# =========================
# CLASS MAPPING
# =========================
names=[
    "rov",
    "plant",
    "animal_fish",
    "animal_starfish",
    "animal_shells",
    "animal_crab",
    "animal_eel",
    "animal_etc",
    "trash_etc",
    "trash_fabric",
    "trash_fishing_gear",
    "trash_metal",
    "trash_paper",
    "trash_plastic",
    "trash_rubber",
    "trash_wood"
]

class_map={
    "rov":0,
    "plant":1,
    "animal_fish":2,
    "animal_starfish":3,
    "animal_shells":4,
    "animal_crab":5,
    "animal_eel":6,
    "animal_etc":7,
    "trash_etc":8,
    "trash_fabric":9,
    "trash_fishing_gear":10,
    "trash_metal":11,
    "trash_paper":12,
    "trash_plastic":13,
    "trash_rubber":14,
    "trash_wood":15
}

# =========================
# CLEAN OUTPUT
# =========================
if os.path.exists(OUT):
    shutil.rmtree(OUT)

for split in ["train","val","test"]:
    os.makedirs(os.path.join(OUT,"images",split),exist_ok=True)
    os.makedirs(os.path.join(OUT,"labels",split),exist_ok=True)

# =========================
# GET IMAGE HASH
# =========================
def sha256(path):
    h=hashlib.sha256()
    with open(path,"rb") as f:
        while True:
            chunk=f.read(1024*1024)
            if not chunk:
                break
            h.update(chunk)
    return h.hexdigest()

# =========================
# LOAD ORIGINAL TRAIN POOL
# =========================
train_files=sorted([
    f for f in os.listdir(TRAIN_SRC)
    if f.lower().endswith((".jpg",".jpeg",".png"))
])

val_files=sorted([
    f for f in os.listdir(VAL_SRC)
    if f.lower().endswith((".jpg",".jpeg",".png"))
])

print("Original train pool:",len(train_files))
print("Original val pool:",len(val_files))

assert len(train_files)==6008
assert len(val_files)==1204

# =========================
# SHA-256 GROUPING
# =========================
hash_groups=defaultdict(list)

for i,f in enumerate(train_files):
    p=os.path.join(IMG_SRC,f)
    h=sha256(p)
    hash_groups[h].append(f)

groups=list(hash_groups.values())

print("Unique SHA-256 groups:",len(groups))
print("Duplicate groups:",sum(len(g)>1 for g in groups))
print("Largest duplicate group:",max(len(g) for g in groups))

# =========================
# GROUP-AWARE 80/20 SPLIT
# TARGET TEST = 1202
# =========================
random.seed(42)

random.shuffle(groups)

target_test=1202
test_groups=[]
test_count=0

# First pass: choose whole groups without exceeding target
remaining=[]

for g in groups:
    if test_count+len(g)<=target_test:
        test_groups.append(g)
        test_count+=len(g)
    else:
        remaining.append(g)

# If exact target was not reached, try to construct the remaining
# amount from unused groups.
need=target_test-test_count

if need>0:
    candidates=sorted(remaining,key=len)

    used=set()

    # Try combinations of small groups to hit exact target
    for g in candidates:
        if len(g)<=need:
            test_groups.append(g)
            test_count+=len(g)
            need-=len(g)
            used.add(id(g))
            if need==0:
                break

if test_count!=target_test:
    raise RuntimeError(
        f"Could not create exact 1202-image group-safe test split. "
        f"Got {test_count}. Do NOT continue; inspect duplicate groups."
    )

test_files=set(f for g in test_groups for f in g)
train_final=[f for f in train_files if f not in test_files]

print("\nV2 split:")
print("Train:",len(train_final))
print("Test :",len(test_files))
print("Val  :",len(val_files))

assert len(train_final)==4806
assert len(test_files)==1202
assert len(val_files)==1204

# =========================
# IMAGE COPY
# =========================
def copy_images(files,split):
    for f in files:
        src=os.path.join(IMG_SRC,f)
        dst=os.path.join(OUT,"images",split,f)
        shutil.copy2(src,dst)

copy_images(train_final,"train")
copy_images(val_files,"val")
copy_images(list(test_files),"test")

print("\nImages copied.")

# =========================
# SUPERVISELY BITMAP -> BBOX
# =========================
def bitmap_bbox(obj):
    bitmap=obj["bitmap"]

    compressed=base64.b64decode(bitmap["data"])
    png_bytes=zlib.decompress(compressed)

    mask=Image.open(io.BytesIO(png_bytes)).convert("L")

    # Supervisely bitmap origin
    ox,oy=bitmap["origin"]

    pixels=mask.load()
    xs=[]
    ys=[]

    for y in range(mask.height):
        for x in range(mask.width):
            if pixels[x,y]>0:
                xs.append(x)
                ys.append(y)

    if not xs:
        return None

    x1=ox+min(xs)
    y1=oy+min(ys)
    x2=ox+max(xs)+1
    y2=oy+max(ys)+1

    return x1,y1,x2,y2

# =========================
# CONVERT JSON -> YOLO
# =========================
def convert_labels(files,split):
    class_counter=Counter()
    missing_ann=0
    empty_ann=0
    invalid=0

    for f in files:
        ann_file=f+".json"
        ann_path=os.path.join(ANN_SRC,ann_file)

        if not os.path.exists(ann_path):
            missing_ann+=1
            continue

        with open(ann_path,"r") as fp:
            data=json.load(fp)

        width=data["size"]["width"]
        height=data["size"]["height"]

        yolo=[]

        for obj in data.get("objects",[]):

            # We only use trash/animal/plant/rov material classes
            if obj.get("classTitle")!="trash":
                continue

            material=None

            for tag in obj.get("tags",[]):
                if tag.get("name")=="material":
                    material=tag.get("value")
                    break

            instance=None

            for tag in obj.get("tags",[]):
                if tag.get("name")=="instance":
                    instance=tag.get("value")
                    break

            # Build the material_version class name
            if material=="rov":
                cls_name="rov"
            elif material=="plant":
                cls_name="plant"
            elif material=="animal":
                animal_map={
                    "fish":"animal_fish",
                    "starfish":"animal_starfish",
                    "shells":"animal_shells",
                    "crab":"animal_crab",
                    "eel":"animal_eel"
                }
                cls_name=animal_map.get(instance,"animal_etc")
            else:
                trash_map={
                    "fabric":"trash_fabric",
                    "fishing_gear":"trash_fishing_gear",
                    "metal":"trash_metal",
                    "paper":"trash_paper",
                    "plastic":"trash_plastic",
                    "rubber":"trash_rubber",
                    "wood":"trash_wood",
                    "etc":"trash_etc"
                }
                cls_name=trash_map.get(material,"trash_etc")

            if cls_name not in class_map:
                invalid+=1
                continue

            bbox=bitmap_bbox(obj)

            if bbox is None:
                invalid+=1
                continue

            x1,y1,x2,y2=bbox

            # Clamp
            x1=max(0,min(x1,width))
            y1=max(0,min(y1,height))
            x2=max(0,min(x2,width))
            y2=max(0,min(y2,height))

            bw=x2-x1
            bh=y2-y1

            if bw<=0 or bh<=0:
                invalid+=1
                continue

            xc=(x1+x2)/2
            yc=(y1+y2)/2

            cls=class_map[cls_name]

            yolo.append(
                f"{cls} {xc/width:.6f} {yc/height:.6f} "
                f"{bw/width:.6f} {bh/height:.6f}"
            )

            class_counter[cls_name]+=1

        label_path=os.path.join(
            OUT,"labels",split,
            os.path.splitext(f)[0]+".txt"
        )

        with open(label_path,"w") as fp:
            fp.write("\n".join(yolo))

        if len(yolo)==0:
            empty_ann+=1

    return class_counter,missing_ann,empty_ann,invalid

# =========================
# CONVERT ALL THREE SPLITS
# =========================
print("\nConverting annotations...")

for split,files in [
    ("train",train_final),
    ("val",val_files),
    ("test",list(test_files))
]:

    counts,missing,empty,invalid=convert_labels(files,split)

    print(f"\n{split}:")
    print("images:",len(files))
    print("missing annotations:",missing)
    print("empty annotations:",empty)
    print("invalid objects:",invalid)
    print("class counts:")

    for name in names:
        print(f"  {name}: {counts[name]}")

# =========================
# WRITE YAML
# =========================
yaml_path="/kaggle/working/data_v2.yaml"

with open(yaml_path,"w") as f:
    f.write(f"path: {OUT}\n")
    f.write("train: images/train\n")
    f.write("val: images/val\n")
    f.write("test: images/test\n")
    f.write("nc: 16\n")
    f.write("names:\n")

    for name in names:
        f.write(f"- {name}\n")

print("\nYAML created:",yaml_path)

# =========================
# FINAL STRUCTURE CHECK
# =========================
for split in ["train","val","test"]:
    ni=len(os.listdir(os.path.join(OUT,"images",split)))
    nl=len(os.listdir(os.path.join(OUT,"labels",split)))

    print(f"{split}: images={ni}, labels={nl}")

print("\nV2 DATASET CREATION COMPLETE.")
```

```python
import os
import json
from collections import Counter

ANN_SRC="/kaggle/input/datasets/hanishkavya/material-version/dataset/original_data/annotations"

class_titles=Counter()
material_tags=Counter()
instance_tags=Counter()
pairs=Counter()

for f in os.listdir(ANN_SRC):
    if not f.endswith(".json"):
        continue

    with open(os.path.join(ANN_SRC,f),"r") as fp:
        data=json.load(fp)

    for obj in data.get("objects",[]):
        title=obj.get("classTitle")
        class_titles[title]+=1

        material=None
        instance=None

        for tag in obj.get("tags",[]):
            if tag.get("name")=="material":
                material=tag.get("value")
            if tag.get("name")=="instance":
                instance=tag.get("value")

        material_tags[material]+=1
        instance_tags[instance]+=1
        pairs[(title,material,instance)]+=1

print("CLASS TITLES")
for k,v in class_titles.most_common():
    print(k,":",v)

print("\nMATERIAL TAGS")
for k,v in material_tags.most_common():
    print(k,":",v)

print("\nINSTANCE TAGS")
for k,v in instance_tags.most_common():
    print(k,":",v)

print("\nCLASS TITLE + MATERIAL + INSTANCE")
for k,v in pairs.most_common():
    print(k,":",v)
```

```python
import os,json,shutil,random,hashlib,base64,zlib,cv2
import numpy as np
from collections import Counter
from tqdm import tqdm

BASE="/kaggle/input/datasets/hanishkavya/material-version/dataset"
IMG_SRC=os.path.join(BASE,"original_data/images")
ANN_SRC=os.path.join(BASE,"original_data/annotations")

OUT="/kaggle/working/yolo_v2"
if os.path.exists(OUT):
    shutil.rmtree(OUT)

for split in ["train","val","test"]:
    os.makedirs(os.path.join(OUT,"images",split),exist_ok=True)
    os.makedirs(os.path.join(OUT,"labels",split),exist_ok=True)

names=[
    "rov","plant","animal_fish","animal_starfish","animal_shells",
    "animal_crab","animal_eel","animal_etc","trash_etc",
    "trash_fabric","trash_fishing_gear","trash_metal","trash_paper",
    "trash_plastic","trash_rubber","trash_wood"
]
cls_id={n:i for i,n in enumerate(names)}

def get_class(obj):
    ct=str(obj.get("classTitle",""))

    if ct=="rov":
        return "rov"

    if ct=="unknown":
        return "trash_etc"

    if ct=="bio":
        for tag in obj.get("tags",[]):
            if tag.get("name")=="plant":
                return "plant"

            if tag.get("name")=="animal":
                v=str(tag.get("value","")).strip().lower()

                if v.isdigit():
                    continue

                if v in ["fish"]:
                    return "animal_fish"
                if v in ["starfish","satrfish"]:
                    return "animal_starfish"
                if v in ["shell","shells"]:
                    return "animal_shells"
                if v in ["eel","ee;"]:
                    return "animal_eel"
                if v in ["crab"," crab"]:
                    return "animal_crab"

                return "animal_etc"

        return None

    if ct=="trash":
        material=None

        for tag in obj.get("tags",[]):
            if tag.get("name")=="material":
                material=str(tag.get("value","")).strip().lower()
                break

        if material in ["fabric","fishing_gear","metal","paper","plastic","rubber","wood","etc"]:
            return "trash_"+material

        return "trash_etc"

    return None

def decode_mask(obj):
    raw=zlib.decompress(base64.b64decode(obj["bitmap"]["data"]))
    arr=np.frombuffer(raw,np.uint8)
    mask=cv2.imdecode(arr,cv2.IMREAD_UNCHANGED)
    return mask[:,:,3].astype(np.uint8)

def make_full_mask(obj,h,w):
    sub=decode_mask(obj)
    ox=int(obj["bitmap"]["origin"][0])
    oy=int(obj["bitmap"]["origin"][1])

    mask=np.zeros((h,w),np.uint8)

    sh,sw=sub.shape[:2]
    x2=min(w,ox+sw)
    y2=min(h,oy+sh)

    if ox<w and oy<h and x2>0 and y2>0:
        sx=max(0,-ox)
        sy=max(0,-oy)
        mask[max(0,oy):y2,max(0,ox):x2]=sub[sy:sy+(y2-max(0,oy)),sx:sx+(x2-max(0,ox))]

    return mask

def mask_to_bbox(mask):
    ys,xs=np.where(mask>0)

    if len(xs)==0:
        return None

    x1,x2=xs.min(),xs.max()
    y1,y2=ys.min(),ys.max()

    return x1,y1,x2,y2

# ----------------------------
# Get original material-version pools
# ----------------------------

train_pool=sorted(os.listdir(os.path.join(BASE,"material_version/train")))
val_pool=sorted(os.listdir(os.path.join(BASE,"material_version/val")))

# Use same 80/20 idea as teammate
random.seed(42)
random.shuffle(train_pool)

train_files=train_pool[:4806]
test_files=train_pool[4806:]

print("Train:",len(train_files))
print("Val:",len(val_pool))
print("Test:",len(test_files))

# ----------------------------
# Convert annotations
# ----------------------------

def process(files,split):
    class_counts=Counter()
    empty=0
    missing=0
    total_boxes=0

    for fname in tqdm(files,desc=split):
        ann_name=fname+".json"
        ann_path=os.path.join(ANN_SRC,ann_name)

        if not os.path.exists(ann_path):
            missing+=1
            continue

        with open(ann_path) as f:
            data=json.load(f)

        w=int(data["size"]["width"])
        h=int(data["size"]["height"])
        objects=data.get("objects",[])

        labels=[]

        for obj in objects:
            cls=get_class(obj)

            if cls is None:
                continue

            try:
                mask=make_full_mask(obj,h,w)
                bbox=mask_to_bbox(mask)
            except Exception:
                continue

            if bbox is None:
                continue

            x1,y1,x2,y2=bbox

            bw=x2-x1
            bh=y2-y1

            if bw<=0 or bh<=0:
                continue

            xc=((x1+x2)/2)/w
            yc=((y1+y2)/2)/h
            nw=bw/w
            nh=bh/h

            labels.append(
                f"{cls_id[cls]} {xc:.6f} {yc:.6f} {nw:.6f} {nh:.6f}"
            )

            class_counts[cls]+=1
            total_boxes+=1

        src_img=os.path.join(IMG_SRC,fname)
        dst_img=os.path.join(OUT,"images",split,fname)
        shutil.copy2(src_img,dst_img)

        label_path=os.path.join(
            OUT,"labels",split,
            os.path.splitext(fname)[0]+".txt"
        )

        with open(label_path,"w") as f:
            f.write("\n".join(labels))

        if len(labels)==0:
            empty+=1

    print("\n",split)
    print("images:",len(files))
    print("missing:",missing)
    print("empty:",empty)
    print("boxes:",total_boxes)
    print("classes:")
    for c in names:
        print(f"{c:20s}",class_counts[c])

    return class_counts

train_counts=process(train_files,"train")
val_counts=process(val_pool,"val")
test_counts=process(test_files,"test")

# ----------------------------
# Create YAML
# ----------------------------

yaml_path=os.path.join(OUT,"data.yaml")

with open(yaml_path,"w") as f:
    f.write(f"""path: {OUT}
train: images/train
val: images/val
test: images/test

names:
""")
    for i,n in enumerate(names):
        f.write(f"  {i}: {n}\n")

print("\nDONE")
print("YAML:",yaml_path)
```
