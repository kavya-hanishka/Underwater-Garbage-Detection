```python
from collections import Counter
from pathlib import Path

LABEL_DIR = Path("/kaggle/input/datasets/hanishkavya/trashcan/yolo_dataset/labels/train")

class_counts = Counter()

for file in LABEL_DIR.glob("*.txt"):
    with open(file) as f:
        for line in f:
            if line.strip():
                cls = int(line.split()[0])
                class_counts[cls] += 1

names = [
    "rov","plant","animal_fish","animal_starfish","animal_shells",
    "animal_crab","animal_eel","animal_etc","trash_etc","trash_fabric",
    "trash_fishing_gear","trash_metal","trash_paper","trash_plastic",
    "trash_rubber","trash_wood"
]

for i,name in enumerate(names):
    print(f"{i:2d} {name:20s}: {class_counts[i]}")
```

```python
from pathlib import Path
import shutil
from collections import Counter

SRC = Path("/kaggle/input/datasets/hanishkavya/trashcan/yolo_dataset")
OUT = Path("/kaggle/working/yolo_oversampled")

# Create directories
for split in ["train", "val", "test"]:
    (OUT / "images" / split).mkdir(parents=True, exist_ok=True)
    (OUT / "labels" / split).mkdir(parents=True, exist_ok=True)

# Copy validation and test unchanged
for split in ["val", "test"]:
    for p in (SRC / "images" / split).glob("*"):
        shutil.copy2(p, OUT / "images" / split / p.name)
    for p in (SRC / "labels" / split).glob("*.txt"):
        shutil.copy2(p, OUT / "labels" / split / p.name)

# Thresholds
VERY_RARE = 150
RARE = 250

def get_repeat_factor(label_file):
    classes = set()

    with open(label_file) as f:
        for line in f:
            if line.strip():
                classes.add(int(line.split()[0]))

    # Highest oversampling requirement among classes in this image
    factors = []

    for cls in classes:
        count = class_counts[cls]

        if count < VERY_RARE:
            factors.append(3)
        elif count < RARE:
            factors.append(2)
        else:
            factors.append(1)

    return max(factors, default=1)

# Build oversampled training set
train_label_dir = SRC / "labels" / "train"
train_image_dir = SRC / "images" / "train"

total_copies = 0

for label_path in train_label_dir.glob("*.txt"):

    stem = label_path.stem

    # Find corresponding image
    image_path = None
    for ext in [".jpg", ".jpeg", ".png"]:
        candidate = train_image_dir / (stem + ext)
        if candidate.exists():
            image_path = candidate
            break

    if image_path is None:
        continue

    repeat = get_repeat_factor(label_path)

    for i in range(repeat):
        suffix = "" if i == 0 else f"_copy{i}"

        shutil.copy2(
            image_path,
            OUT / "images" / "train" / f"{stem}{suffix}{image_path.suffix}"
        )

        shutil.copy2(
            label_path,
            OUT / "labels" / "train" / f"{stem}{suffix}.txt"
        )

        total_copies += 1

print("Oversampled training images:", total_copies)
print("Validation images:", len(list((OUT / "images" / "val").glob("*"))))
print("Test images:", len(list((OUT / "images" / "test").glob("*"))))
```
